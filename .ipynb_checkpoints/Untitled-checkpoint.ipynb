```python
import pandas as pd
import numpy as np


# ── Haversine distance (vectorised, returns km) ─────────────────────────
def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
    return R * 2 * np.arcsin(np.sqrt(a))


# ── Find the median (most central) school ───────────────────────────────
def find_median_school(df):
    """
    Returns the row whose total distance to all other schools is minimum.
    Uses the geometric medoid — an actual school, not an imaginary midpoint.
    """
    lats = df["Lattitude"].values
    lons = df["Longitude"].values
    n = len(df)

    total_dist = np.zeros(n)
    for i in range(n):
        total_dist[i] = np.sum(
            haversine(lats[i], lons[i], lats, lons)
        )

    median_idx = np.argmin(total_dist)
    return df.iloc[median_idx], total_dist


# ── Find k nearest schools from a given school ──────────────────────────
def k_nearest_schools(df, school_row, k):
    """
    Given a reference school row, returns the k closest schools
    from the dataframe (excluding the reference school itself).

    Parameters
    ----------
    df         : full schools DataFrame
    school_row : a single row (Series) — e.g. the median school
    k          : number of nearest neighbors to return
    """
    others = df[df["OBJECTID"] != school_row["OBJECTID"]].copy()

    others["distance_km"] = haversine(
        school_row["Lattitude"], school_row["Longitude"],
        others["Lattitude"].values, others["Longitude"].values
    )

    return others.nsmallest(k, "distance_km").reset_index(drop=True)


# ── Usage ────────────────────────────────────────────────────────────────
# df = pd.read_csv("schools.csv")   ← load your data here

k = 4

median_school, all_distances = find_median_school(df)

print("Median (most central) school:")
print(median_school[["Name", "Address", "Longitude", "Lattitude"]])

nearest = k_nearest_schools(df, median_school, k=k)

print(f"\n{k} nearest schools from '{median_school['Name']}':")
print(nearest[["Name", "Address", "distance_km"]].to_string(index=False))
```
