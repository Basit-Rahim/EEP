```python
import pandas as pd

df = pd.read_stata(
    "Lahore.dta",
    convert_categoricals=False
)

df.head()
```

```python
school_children = (
    df[(df["age"] >= 5) & (df["age"] <= 16)]
    .groupby("hhcode")
    .size()
    .rename("school_age_children")
)
```

```python
school_age = df[(df["age"] >= 5) & (df["age"] <= 16)].copy()

school_age["enrolled"] = (
    school_age["education_access"]
    .fillna(0)
    .astype(int)
)

enrollment_rate = school_age["enrolled"].mean()

print(f"Enrollment probability: {enrollment_rate:.2%}")
```
